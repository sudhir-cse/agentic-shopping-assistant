### Import Dependencies

```python
from langsmith import Client
```

### Attach Feedback to a specific trace in LangSmith

```python
client = Client()
```

```python
RUN_ID = "019f538b-3518-72a0-a770-f9abdbd316a2"
```

```python
client.create_feedback(
    run_id=RUN_ID,
    key="feedback",
    score=0,
    feedback_source_type="api"
)
```

```python
client.create_feedback(
    run_id=RUN_ID,
    key="comment",
    value="This is a long form free text feedback left by a user.",
    feedback_source_type="api"
)
```

```python
client.create_feedback(
    run_id=RUN_ID,
    key="comment_with_score",
    value="This is a long form free text feedback left by a user.",
    score=1,
    feedback_source_type="api"
)
```
